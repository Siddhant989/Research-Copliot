````python
def _download_pdf(url):
    """Download PDF to a temp file, return path."""
    suffix = ".pdf"
    tmp = tempfile.NamedTemporaryFile(delete=False, suffix=suffix)
    try:
        urllib.request.urlretrieve(url, tmp.name)
        return tmp.name
    except Exception:
        tmp.close()
        os.unlink(tmp.name)
        raise
````

````python
#for removing the excess space and line change
def _clean(text):

    while "\n\n\n" in text:
        text = text.replace("\n\n\n", "\n\n")

    while "  " in text:
        text = text.replace("  ", " ")

    return text.strip()
````

````python
def _extract_abstract(text):

    lower_text = text.lower()

    start = lower_text.find("abstract")

    if start != -1:

        abstract_text = text[start + len("abstract"):]

        end = abstract_text.lower().find("introduction")

        if end != -1:
            return abstract_text[:end].strip()[:800]

        return abstract_text[:800]

    lines = text.split("\n")

    long_lines = []

    for line in lines:

        if len(line.strip()) > 60:
            long_lines.append(line.strip())

        if len(long_lines) >= 4:
            break

    return " ".join(long_lines)[:800]
````

````python
_SECTION_RE = re.compile(
    r"^(?:\d+\.?\s+)?([A-Z][A-Za-z &\-]{2,50})$",
    re.MULTILINE
)

_COMMON_HEADERS = {
    "abstract",
    "introduction",
    "related work",
    "background",
    "methodology",
    "method",
    "approach",
    "model",
    "architecture",
    "experiments",
    "results",
    "evaluation",
    "discussion",
    "conclusion",
    "references",
    "acknowledgements",
    "appendix"
}


def _extract_sections(raw):

    sections = {}

    current_name = "preamble"

    current_lines = []

    lines = raw.split("\n")

    for line in lines:

        stripped = line.strip()

        is_header = False

        if stripped and len(stripped) < 60:

            if stripped.lower() in _COMMON_HEADERS:
                is_header = True

            elif _SECTION_RE.match(stripped):
                is_header = True

        if is_header:

            if current_lines:

                sections[current_name] = "\n".join(
                    current_lines
                ).strip()

            current_name = stripped.lower()

            current_lines = []

        else:

            current_lines.append(line)

    if current_lines:

        sections[current_name] = "\n".join(
            current_lines
        ).strip()

    return sections
````

````python
def parse_pdf(pdf_path: str) -> dict:
    doc = fitz.open(pdf_path)
    page_count = len(doc)

    pages_text = []
    for i, page in enumerate(doc):
        text = page.get_text("text")
        if text.strip():
            pages_text.append(f"[Page {i}]\n{text}")

    doc.close()

    raw_text = _clean("\n".join(pages_text))
    abstract = _extract_abstract(raw_text)
    sections = _extract_sections(raw_text)

    return {
        "raw_text":   raw_text,
        "abstract":   abstract,
        "sections":   sections,
        "page_count": page_count,
    }
````

````python
def parse_json_response(text):
    text = text.replace("```json", "").replace("```", "").strip()

    # Try direct parse first
    try:
        return json.loads(text)
    except Exception:
        pass

    # If that didn't work, try to find a JSON object somewhere inside the text
    match = re.search(r"\{.*\}", text, re.DOTALL)
    if match:
        try:
            return json.loads(match.group())
        except Exception:
            pass

    # Nothing worked
    return {}
````

````python
import os
import sys
import io
import csv
import json
import base64
import time
import re
from pathlib import Path

import fitz                    # PyMuPDF — page rendering + image extraction
import pdfplumber              # Digital table extraction
from PIL import Image
import requests

# ─────────────────────────────────────────────────────────
# CONFIGURATION
# ─────────────────────────────────────────────────────────
GEMINI_API_KEY  = os.environ.get("GEMINI_API_KEY", "YOUR_API_KEY_HERE")
GEMINI_MODELS   = ["gemini-2.5-flash-lite", "gemini-2.5-flash"]  # tries in order
OUTPUT_DIR      = "extracted_tables"     # all CSVs saved here
RENDER_DPI      = 150                    # DPI for rendering scanned pages to image
MAX_RETRIES     = 3
RETRY_DELAY     = 10
CALL_DELAY      = 3                      # seconds between Gemini calls (rate limit)
# ─────────────────────────────────────────────────────────


def _gemini_request(parts: list, model: str) -> tuple:
    """
    Low-level Gemini call.
    parts = list of {"text": ...} or {"inline_data": {...}} dicts
    Returns (response_text, error_code)
    """
    url = (
        f"https://generativelanguage.googleapis.com/v1beta/models/"
        f"{model}:generateContent?key={GEMINI_API_KEY}"
    )
    payload = {
        "contents": [{"parts": parts}],
        "generationConfig": {"temperature": 0.1, "maxOutputTokens": 1024}
    }
    try:
        r = requests.post(url, json=payload, timeout=60)
        if r.status_code == 200:
            text = r.json()["candidates"][0]["content"]["parts"][0]["text"].strip()
            return text, None
        elif r.status_code == 429:  return None, "RATE_LIMIT"
        elif r.status_code == 503:  return None, "OVERLOADED"
        elif r.status_code == 404:  return None, "MODEL_NOT_FOUND"
        else:                       return None, f"HTTP_{r.status_code}"
    except requests.exceptions.Timeout:
        return None, "TIMEOUT"
    except Exception as e:
        return None, str(e)


def gemini_call(parts: list) -> str:
    """Retry + model-fallback wrapper around _gemini_request."""
    for model in GEMINI_MODELS:
        for attempt in range(1, MAX_RETRIES + 1):
            print(f"       [Gemini] model={model} attempt={attempt} …", end=" ")
            text, err = _gemini_request(parts, model)
            if text:
                print("✅")
                return text
            print(f"❌ ({err})")
            if err == "MODEL_NOT_FOUND":
                break
            wait = RETRY_DELAY * (2 if err == "RATE_LIMIT" else 1)
            print(f"       ⏳ waiting {wait}s …")
            time.sleep(wait)
    return "❌ Gemini could not respond after all retries."


# ══════════════════════════════════════════════════════════
# CSV UTILITIES — make tables Streamlit-ready
# ══════════════════════════════════════════════════════════

def save_csv(rows: list[list], filepath: str):
    """
    Save a list-of-lists table to CSV.
    - Strips extra whitespace from every cell
    - Ensures the first row becomes a clean header (no blank headers)
    - Removes fully empty rows
    """
    if not rows:
        return

    # Clean every cell
    cleaned = []
    for row in rows:
        clean_row = [str(cell).strip() for cell in row]
        if any(cell for cell in clean_row):   # skip fully blank rows
            cleaned.append(clean_row)

    if not cleaned:
        return

    # Fix blank header cells: "Col_1", "Col_2" …
    header = cleaned[0]
    header = [h if h else f"Col_{i+1}" for i, h in enumerate(header)]
    cleaned[0] = header

    os.makedirs(os.path.dirname(filepath), exist_ok=True)
    with open(filepath, "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerows(cleaned)


def image_to_b64(pil_img: Image.Image) -> str:
    """Convert a PIL image to base64 PNG string."""
    buf = io.BytesIO()
    pil_img.save(buf, format="PNG")
    return base64.b64encode(buf.getvalue()).decode("utf-8")


# ══════════════════════════════════════════════════════════
# PATH 1 — DIGITAL TABLE EXTRACTION (pdfplumber)
# ══════════════════════════════════════════════════════════

def extract_digital_tables(pdf_path: str) -> list[dict]:
    """
    Use pdfplumber to find and extract text-based tables.
    Returns list of table records.
    """
    results = []
    with pdfplumber.open(pdf_path) as pdf:
        for page_num, page in enumerate(pdf.pages, start=1):
            tables = page.extract_tables()
            for tbl_idx, table in enumerate(tables, start=1):
                if not table or len(table) < 2:
                    continue   # skip single-row noise

                label    = f"digital_page{page_num}_table{tbl_idx}"
                csv_path = os.path.join(OUTPUT_DIR, f"{label}.csv")
                save_csv(table, csv_path)

                print(f"  ✅ [Digital] Page {page_num}, Table {tbl_idx} "
                      f"→ {label}.csv  ({len(table)} rows × {len(table[0])} cols)")

                results.append({
                    "label":    label,
                    "page":     page_num,
                    "index":    tbl_idx,
                    "type":     "digital",
                    "csv_path": csv_path,
                    "rows":     table,
                })
    return results


# ══════════════════════════════════════════════════════════
# PATH 2 — SCANNED TABLE EXTRACTION (PyMuPDF → Gemini Vision)
# ══════════════════════════════════════════════════════════

def has_digital_tables_on_page(pdf_path: str, page_num: int) -> bool:
    """Returns True if pdfplumber already found tables on this page (1-indexed)."""
    with pdfplumber.open(pdf_path) as pdf:
        page = pdf.pages[page_num - 1]
        tables = page.extract_tables()
        return len(tables) > 0


def render_page_as_image(pdf_path: str, page_num: int) -> Image.Image:
    """Render a PDF page (1-indexed) to a PIL image at RENDER_DPI."""
    doc  = fitz.open(pdf_path)
    page = doc[page_num - 1]
    mat  = fitz.Matrix(RENDER_DPI / 72, RENDER_DPI / 72)
    pix  = page.get_pixmap(matrix=mat, alpha=False)
    img  = Image.frombytes("RGB", [pix.width, pix.height], pix.samples)
    doc.close()
    return img


def ask_gemini_for_scanned_table_csv(page_img: Image.Image, page_num: int, tbl_idx: int) -> str | None:
    """
    Send a page image to Gemini and ask it to return the table as CSV text.
    Returns raw CSV string, or None if no table found.
    """
    b64 = image_to_b64(page_img)
    parts = [
        {"inline_data": {"mime_type": "image/png", "data": b64}},
        {"text": (
            f"This is page {page_num} of a research paper (scanned image). "
            f"Look for table #{tbl_idx} on this page. "
            "If a table exists, extract ALL its data and return it ONLY as raw CSV "
            "(comma-separated values, one row per line, first row = header). "
            "Do NOT add any explanation, markdown fences, or extra text. "
            "If no table is present, reply with exactly: NO_TABLE"
        )}
    ]
    response = gemini_call(parts)
    response = response.strip()

    # Remove markdown code fences if Gemini added them anyway
    response = re.sub(r"^```[a-z]*\n?", "", response, flags=re.IGNORECASE)
    response = re.sub(r"\n?```$", "", response)
    response = response.strip()

    if response.upper() == "NO_TABLE" or not response:
        return None
    return response


def csv_text_to_rows(csv_text: str) -> list[list]:
    """Parse a raw CSV string into a list-of-lists."""
    reader = csv.reader(io.StringIO(csv_text))
    return [row for row in reader if any(cell.strip() for cell in row)]


def extract_scanned_tables(pdf_path: str, digital_pages: set) -> list[dict]:
    """
    For pages that pdfplumber found NO tables on, render the page as an image
    and ask Gemini Vision to extract any tables it sees.
    """
    doc        = fitz.open(pdf_path)
    total_pages = len(doc)
    doc.close()

    results = []

    for page_num in range(1, total_pages + 1):
        # Skip pages that already had digital tables
        if page_num in digital_pages:
            continue

        print(f"  🔍 [Scan check] Page {page_num} …", end=" ")
        page_img = render_page_as_image(pdf_path, page_num)

        # Ask Gemini if there's a table on this page (table index 1 first)
        tbl_idx = 1
        while True:
            print(f"asking Gemini for table {tbl_idx} …")
            csv_text = ask_gemini_for_scanned_table_csv(page_img, page_num, tbl_idx)
            time.sleep(CALL_DELAY)

            if csv_text is None:
                # No more tables on this page
                if tbl_idx == 1:
                    print(f"       No table found on page {page_num}.")
                break

            rows     = csv_text_to_rows(csv_text)
            if len(rows) < 2:
                break

            label    = f"scanned_page{page_num}_table{tbl_idx}"
            csv_path = os.path.join(OUTPUT_DIR, f"{label}.csv")
            save_csv(rows, csv_path)

            print(f"  ✅ [Scanned] Page {page_num}, Table {tbl_idx} "
                  f"→ {label}.csv  ({len(rows)} rows × {len(rows[0])} cols)")

            results.append({
                "label":    label,
                "page":     page_num,
                "index":    tbl_idx,
                "type":     "scanned",
                "csv_path": csv_path,
                "rows":     rows,
            })

            tbl_idx += 1
            # Ask Gemini if there's another table on the same page
            # (stop after 5 to avoid infinite loops on complex pages)
            if tbl_idx > 5:
                break

    return results


# ══════════════════════════════════════════════════════════
# STEP 3 — GET EXPLANATION FOR EVERY TABLE
# ══════════════════════════════════════════════════════════

def explain_table(table: dict) -> str:
    """
    Ask Gemini to explain a table.
    For digital tables → send the raw CSV text (no image needed, saves quota).
    For scanned tables → we already have the CSV rows, same approach.
    """
    # Build a compact text representation of the table
    rows = table["rows"]
    if len(rows) > 30:
        # For very large tables, send header + first 20 rows + note
        sample_rows = rows[:21]
        note = f"\n[Table truncated — showing 20 of {len(rows)-1} data rows for context]"
    else:
        sample_rows = rows
        note = ""

    buf = io.StringIO()
    csv.writer(buf).writerows(sample_rows)
    csv_preview = buf.getvalue() + note

    parts = [{"text": (
        "You are analyzing a table from a research paper.\n"
        "Here is the table data in CSV format:\n\n"
        f"{csv_preview}\n\n"
        "Please explain this table clearly and concisely:\n"
        "1. What is the table about? (its purpose / topic)\n"
        "2. What do the rows and columns represent?\n"
        "3. What are the key values, trends, or comparisons visible?\n"
        "4. What is the main takeaway or conclusion from this table?\n"
        "Keep the explanation under 150 words."
    )}]
    return gemini_call(parts)


# ══════════════════════════════════════════════════════════
# STEP 4 — SAVE MARKDOWN REPORT
# ══════════════════════════════════════════════════════════

def save_report(all_tables: list[dict], pdf_name: str):
    """
    Save a Markdown report with:
    - Path to each CSV (copy-paste ready for st.dataframe)
    - Gemini's explanation of each table
    """
    report_path = os.path.join(OUTPUT_DIR, "table_explanations.md")

    with open(report_path, "w", encoding="utf-8") as f:
        f.write(f"# Table Explanations — {pdf_name}\n\n")
        f.write(
            "> **How to use in Streamlit:**\n"
            "> ```python\n"
            "> import pandas as pd, streamlit as st\n"
            "> df = pd.read_csv('extracted_tables/<filename>.csv')\n"
            "> st.dataframe(df, use_container_width=True)\n"
            "> ```\n\n---\n\n"
        )

        for t in all_tables:
            f.write(f"## {t['label']}\n\n")
            f.write(f"| Field | Value |\n|---|---|\n")
            f.write(f"| **Page** | {t['page']} |\n")
            f.write(f"| **Type** | {t['type'].capitalize()} PDF |\n")
            f.write(f"| **CSV file** | `{t['csv_path']}` |\n")
            f.write(f"| **Rows** | {len(t['rows'])} |\n")
            f.write(f"| **Columns** | {len(t['rows'][0]) if t['rows'] else 'N/A'} |\n\n")
            f.write(f"### Explanation\n\n{t['explanation']}\n\n---\n\n")

    print(f"\n📄  Report saved → {report_path}")
    return report_path


# ══════════════════════════════════════════════════════════
# MAIN
# ══════════════════════════════════════════════════════════

def main():
    if len(sys.argv) < 2:
        print("Usage:  python pdf_table_extractor.py <path_to_pdf>")
        sys.exit(1)

    pdf_path = sys.argv[1]
    if not os.path.exists(pdf_path):
        print(f"❌  File not found: {pdf_path}")
        sys.exit(1)

    if GEMINI_API_KEY == "YOUR_API_KEY_HERE":
        print("❌  Set your GEMINI_API_KEY first.")
        print("    export GEMINI_API_KEY='your_key_here'")
        print("    Get a free key → https://aistudio.google.com/app/apikey")
        sys.exit(1)

    os.makedirs(OUTPUT_DIR, exist_ok=True)

    print(f"\n{'═'*55}")
    print(f"  PDF Table Extractor + Gemini Explainer")
    print(f"{'═'*55}")
    print(f"  File : {pdf_path}\n")

    # ── STEP 1 : Digital tables (pdfplumber) ───────────
    print("📊  STEP 1 — Extracting digital tables (pdfplumber) …\n")
    digital_tables = extract_digital_tables(pdf_path)
    digital_pages  = {t["page"] for t in digital_tables}
    print(f"\n  → Found {len(digital_tables)} digital table(s).\n")

    # ── STEP 2 : Scanned tables (Gemini Vision) ────────
    print("🖼️   STEP 2 — Checking remaining pages for scanned tables …\n")
    scanned_tables = extract_scanned_tables(pdf_path, digital_pages)
    print(f"\n  → Found {len(scanned_tables)} scanned table(s).\n")

    all_tables = digital_tables + scanned_tables

    if not all_tables:
        print("⚠️   No tables found in this PDF.")
        return

    # ── STEP 3 : Explain every table ──────────────────
    print(f"{'─'*55}")
    print(f"🤖  STEP 3 — Explaining {len(all_tables)} table(s) via Gemini …\n")

    for i, table in enumerate(all_tables):
        print(f"\n[{i+1}/{len(all_tables)}]  {table['label']}")
        explanation = explain_table(table)
        table["explanation"] = explanation
        print(f"     📝  {explanation[:120]}{'…' if len(explanation)>120 else ''}")
        if i < len(all_tables) - 1:
            time.sleep(CALL_DELAY)

    # ── STEP 4 : Save report ───────────────────────────
    report_path = save_report(all_tables, Path(pdf_path).name)

    print(f"\n{'═'*55}")
    print(f"✅  Done!")
    print(f"   CSVs    → ./{OUTPUT_DIR}/*.csv")
    print(f"   Report  → {report_path}")
    print(f"\n   Streamlit usage:")
    print(f"   import pandas as pd, streamlit as st")
    print(f"   df = pd.read_csv('{OUTPUT_DIR}/<filename>.csv')")
    print(f"   st.dataframe(df, use_container_width=True)")
    print(f"{'═'*55}\n")


if __name__ == "__main__":
    main()
````

````python
def _gemini_request(parts: list, model: str) -> tuple:
    """
    Low-level Gemini call.
    parts = list of {"text": ...} or {"inline_data": {...}} dicts
    Returns (response_text, error_code)
    """
    url = (
        f"https://generativelanguage.googleapis.com/v1beta/models/"
        f"{model}:generateContent?key={GEMINI_API_KEY}"
    )
    payload = {
        "contents": [{"parts": parts}],
        "generationConfig": {"temperature": 0.1, "maxOutputTokens": 1024}
    }
    try:
        r = requests.post(url, json=payload, timeout=60)
        if r.status_code == 200:
            text = r.json()["candidates"][0]["content"]["parts"][0]["text"].strip()
            return text, None
        elif r.status_code == 429:  return None, "RATE_LIMIT"
        elif r.status_code == 503:  return None, "OVERLOADED"
        elif r.status_code == 404:  return None, "MODEL_NOT_FOUND"
        else:                       return None, f"HTTP_{r.status_code}"
    except requests.exceptions.Timeout:
        return None, "TIMEOUT"
    except Exception as e:
        return None, str(e)


def gemini_call(parts: list) -> str:
    """Retry + model-fallback wrapper around _gemini_request."""
    for model in GEMINI_MODELS:
        for attempt in range(1, MAX_RETRIES + 1):
            print(f"       [Gemini] model={model} attempt={attempt} …", end=" ")
            text, err = _gemini_request(parts, model)
            if text:
                print("✅")
                return text
            print(f"❌ ({err})")
            if err == "MODEL_NOT_FOUND":
                break
            wait = RETRY_DELAY * (2 if err == "RATE_LIMIT" else 1)
            print(f"       ⏳ waiting {wait}s …")
            time.sleep(wait)
    return "❌ Gemini could not respond after all retries."


# ══════════════════════════════════════════════════════════
# CSV UTILITIES — make tables Streamlit-ready
# ══════════════════════════════════════════════════════════

def save_csv(rows: list[list], filepath: str):
    """
    Save a list-of-lists table to CSV.
    - Strips extra whitespace from every cell
    - Ensures the first row becomes a clean header (no blank headers)
    - Removes fully empty rows
    """
    if not rows:
        return

    # Clean every cell
    cleaned = []
    for row in rows:
        clean_row = [str(cell).strip() for cell in row]
        if any(cell for cell in clean_row):   # skip fully blank rows
            cleaned.append(clean_row)

    if not cleaned:
        return

    # Fix blank header cells: "Col_1", "Col_2" …
    header = cleaned[0]
    header = [h if h else f"Col_{i+1}" for i, h in enumerate(header)]
    cleaned[0] = header

    os.makedirs(os.path.dirname(filepath), exist_ok=True)
    with open(filepath, "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerows(cleaned)


def image_to_b64(pil_img: Image.Image) -> str:
    """Convert a PIL image to base64 PNG string."""
    buf = io.BytesIO()
    pil_img.save(buf, format="PNG")
    return base64.b64encode(buf.getvalue()).decode("utf-8")


# ══════════════════════════════════════════════════════════
# PATH 1 — DIGITAL TABLE EXTRACTION (pdfplumber)
# ══════════════════════════════════════════════════════════

def extract_digital_tables(pdf_path: str) -> list[dict]:
    """
    Use pdfplumber to find and extract text-based tables.
    Returns list of table records.
    """
    results = []
    with pdfplumber.open(pdf_path) as pdf:
        for page_num, page in enumerate(pdf.pages, start=1):
            tables = page.extract_tables()
            for tbl_idx, table in enumerate(tables, start=1):
                if not table or len(table) < 2:
                    continue   # skip single-row noise

                label    = f"digital_page{page_num}_table{tbl_idx}"
                csv_path = os.path.join(OUTPUT_DIR, f"{label}.csv")
                save_csv(table, csv_path)

                print(f"  ✅ [Digital] Page {page_num}, Table {tbl_idx} "
                      f"→ {label}.csv  ({len(table)} rows × {len(table[0])} cols)")

                results.append({
                    "label":    label,
                    "page":     page_num,
                    "index":    tbl_idx,
                    "type":     "digital",
                    "csv_path": csv_path,
                    "rows":     table,
                })
    return results


# ══════════════════════════════════════════════════════════
# PATH 2 — SCANNED TABLE EXTRACTION (PyMuPDF → Gemini Vision)
# ══════════════════════════════════════════════════════════

def has_digital_tables_on_page(pdf_path: str, page_num: int) -> bool:
    """Returns True if pdfplumber already found tables on this page (1-indexed)."""
    with pdfplumber.open(pdf_path) as pdf:
        page = pdf.pages[page_num - 1]
        tables = page.extract_tables()
        return len(tables) > 0


def render_page_as_image(pdf_path: str, page_num: int) -> Image.Image:
    """Render a PDF page (1-indexed) to a PIL image at RENDER_DPI."""
    doc  = fitz.open(pdf_path)
    page = doc[page_num - 1]
    mat  = fitz.Matrix(RENDER_DPI / 72, RENDER_DPI / 72)
    pix  = page.get_pixmap(matrix=mat, alpha=False)
    img  = Image.frombytes("RGB", [pix.width, pix.height], pix.samples)
    doc.close()
    return img


def ask_gemini_for_scanned_table_csv(page_img: Image.Image, page_num: int, tbl_idx: int) -> str | None:
    """
    Send a page image to Gemini and ask it to return the table as CSV text.
    Returns raw CSV string, or None if no table found.
    """
    b64 = image_to_b64(page_img)
    parts = [
        {"inline_data": {"mime_type": "image/png", "data": b64}},
        {"text": (
            f"This is page {page_num} of a research paper (scanned image). "
            f"Look for table #{tbl_idx} on this page. "
            "If a table exists, extract ALL its data and return it ONLY as raw CSV "
            "(comma-separated values, one row per line, first row = header). "
            "Do NOT add any explanation, markdown fences, or extra text. "
            "If no table is present, reply with exactly: NO_TABLE"
        )}
    ]
    response = gemini_call(parts)
    response = response.strip()

    # Remove markdown code fences if Gemini added them anyway
    response = re.sub(r"^```[a-z]*\n?", "", response, flags=re.IGNORECASE)
    response = re.sub(r"\n?```$", "", response)
    response = response.strip()

    if response.upper() == "NO_TABLE" or not response:
        return None
    return response


def csv_text_to_rows(csv_text: str) -> list[list]:
    """Parse a raw CSV string into a list-of-lists."""
    reader = csv.reader(io.StringIO(csv_text))
    return [row for row in reader if any(cell.strip() for cell in row)]
````

````python
import re
import fitz  # PyMuPDF

def find_table_pages(pdf_path):
    """
    Returns a sorted list of pages containing table captions
    such as:
        Table 1
        TABLE 2
        Table 3:
        Table 4.
    """

    doc = fitz.open(pdf_path)

    table_pages = set()

    # Matches:
    # Table 1
    # TABLE 2:
    # Table 3.
    # Table 10 Results
    pattern = re.compile(
        r'\btable\s+\d+\b',
        re.IGNORECASE
    )

    for page_num in range(len(doc)):
        page = doc[page_num]
        text = page.get_text("text")

        matches = pattern.findall(text)

        if matches:
            table_pages.add(page_num + 1)  # 1-based page numbering

    doc.close()

    return sorted(table_pages)


# Example


pages = find_table_pages(pdf_path)

print("Table pages:", pages)
````

````python
def extract_scanned_tables(pdf_path: str, digital_pages: set) -> list[dict]:
    """
    For pages that pdfplumber found NO tables on, render the page as an image
    and ask Gemini Vision to extract any tables it sees.
    """
    doc        = fitz.open(pdf_path)
    doc.close()

    results = []

    for page_num in digital_pages:
        print(f"  🔍 [Scan check] Page {page_num} …", end=" ")
        page_img = render_page_as_image(pdf_path, page_num)

        # Ask Gemini if there's a table on this page (table index 1 first)
        tbl_idx = 1
        while True:
            print(f"asking Gemini for table {tbl_idx} …")
            csv_text = ask_gemini_for_scanned_table_csv(page_img, page_num, tbl_idx)
            time.sleep(CALL_DELAY)

            if csv_text is None:
                # No more tables on this page
                if tbl_idx == 1:
                    print(f"       No table found on page {page_num}.")
                break

            rows     = csv_text_to_rows(csv_text)
            if len(rows) < 2:
                break

            label    = f"scanned_page{page_num}_table{tbl_idx}"
            csv_path = os.path.join(OUTPUT_DIR, f"{label}.csv")
            save_csv(rows, csv_path)

            print(f"  ✅ [Scanned] Page {page_num}, Table {tbl_idx} "
                  f"→ {label}.csv  ({len(rows)} rows × {len(rows[0])} cols)")

            results.append({
                "label":    label,
                "page":     page_num,
                "index":    tbl_idx,
                "type":     "scanned",
                "csv_path": csv_path,
                "rows":     rows,
            })

            tbl_idx += 1
            # Ask Gemini if there's another table on the same page
            # (stop after 5 to avoid infinite loops on complex pages)
            if tbl_idx > 5:
                break

    return results
````

````python
import os
import sys
import io
import re
import csv
import base64
import time
from pathlib import Path

import fitz
from PIL import Image
import requests

# ─────────────────────────────────────────────────────
# CONFIGURATION
# ─────────────────────────────────────────────────────
GEMINI_MODELS  = ["gemini-2.5-flash-lite", "gemini-2.5-flash"]
OUTPUT_DIR     = "extracted_tables"
RENDER_DPI     = 150
MAX_RETRIES    = 3
RETRY_DELAY    = 10
CALL_DELAY     = 3
# ─────────────────────────────────────────────────────


# ══════════════════════════════════════════════════════
# GEMINI HELPERS
# ══════════════════════════════════════════════════════

def image_to_b64(pil_img: Image.Image) -> str:
    buf = io.BytesIO()
    pil_img.save(buf, format="PNG")
    return base64.b64encode(buf.getvalue()).decode("utf-8")


def render_page(pdf_path: str, page_index: int) -> Image.Image:
    """Render a PDF page (0-indexed) to PIL image."""
    doc  = fitz.open(pdf_path)
    page = doc[page_index]
    mat  = fitz.Matrix(RENDER_DPI / 72, RENDER_DPI / 72)
    pix  = page.get_pixmap(matrix=mat, alpha=False)
    img  = Image.frombytes("RGB", [pix.width, pix.height], pix.samples)
    doc.close()
    return img


def _gemini_request(parts: list, model: str) -> tuple:
    """Raw Gemini API call. Returns (text, error)."""
    url = (
        f"https://generativelanguage.googleapis.com/v1beta/models/"
        f"{model}:generateContent?key={GEMINI_API_KEY}"
    )
    payload = {
        "contents": [{"parts": parts}],
        "generationConfig": {"temperature": 0.1, "maxOutputTokens": 1024}
    }
    try:
        r = requests.post(url, json=payload, timeout=60)
        if r.status_code == 200:
            text = r.json()["candidates"][0]["content"]["parts"][0]["text"].strip()
            return text, None
        elif r.status_code == 429: return None, "RATE_LIMIT"
        elif r.status_code == 503: return None, "OVERLOADED"
        elif r.status_code == 404: return None, "MODEL_NOT_FOUND"
        else:                      return None, f"HTTP_{r.status_code}"
    except requests.exceptions.Timeout:
        return None, "TIMEOUT"
    except Exception as e:
        return None, str(e)


def gemini_call(parts: list) -> str:
    """Model fallback + retry wrapper."""
    for model in GEMINI_MODELS:
        for attempt in range(1, MAX_RETRIES + 1):
            print(f"      [Gemini] {model} attempt {attempt} … ", end="", flush=True)
            text, err = _gemini_request(parts, model)
            if text:
                print("✅")
                return text
            print(f"❌ ({err})")
            if err == "MODEL_NOT_FOUND":
                break
            wait = RETRY_DELAY * (2 if err == "RATE_LIMIT" else 1)
            print(f"      ⏳ waiting {wait}s …")
            time.sleep(wait)
    return None


# ══════════════════════════════════════════════════════
# STEP 1 — REGEX: find pages with table captions
# ══════════════════════════════════════════════════════

def find_table_pages_regex(pdf_path: str) -> list:
    """
    Returns sorted list of 1-indexed pages containing captions like
    'Table 1', 'TABLE 2:', 'Table 3.' etc.
    """
    doc     = fitz.open(pdf_path)
    pattern = re.compile(r'\btable\s+\d+\b', re.IGNORECASE)
    pages   = set()

    for page_num in range(len(doc)):
        text    = doc[page_num].get_text("text")
        matches = pattern.findall(text)
        if matches:
            pages.add(page_num + 1)   # 1-based

    doc.close()
    return sorted(pages)


# ══════════════════════════════════════════════════════
# STEP 2 — GEMINI VISION: extract table from a page
# ══════════════════════════════════════════════════════

def extract_table_from_page(pdf_path: str, page_num: int) -> tuple:
    """
    Send page image to Gemini. Ask it to return the table as raw CSV.
    Returns (csv_text, explanation) or (None, None) if no table found.
    """
    img = render_page(pdf_path, page_num - 1)   # convert to 0-indexed
    b64 = image_to_b64(img)

    # ── Extract CSV ──────────────────────────────────
    csv_parts = [
        {"inline_data": {"mime_type": "image/png", "data": b64}},
        {"text": (
            f"This is page {page_num} of a research paper. "
            "Find the table on this page and extract ALL its data. "
            "Return it ONLY as raw CSV (comma-separated, one row per line, "
            "first row = header). "
            "No explanation, no markdown fences, no extra text. "
            "If no table exists reply exactly: NO_TABLE"
        )}
    ]
    csv_response = gemini_call(csv_parts)
    time.sleep(CALL_DELAY)

    if csv_response is None or csv_response.strip().upper() == "NO_TABLE":
        return None, None

    # Clean markdown fences if Gemini added them
    csv_text = re.sub(r"^```[a-z]*\n?", "", csv_response, flags=re.IGNORECASE)
    csv_text = re.sub(r"\n?```$", "", csv_text).strip()

    if not csv_text:
        return None, None

    # ── Explain the table ────────────────────────────
    explain_parts = [
        {"text": (
            "You are analyzing a table from a research paper.\n"
            "Here is the table in CSV format:\n\n"
            f"{csv_text}\n\n"
            "Explain this table concisely:\n"
            "1. What is the table about?\n"
            "2. What do the rows and columns represent?\n"
            "3. Key values, trends, or comparisons?\n"
            "4. Main takeaway?\n"
            "Keep it under 150 words."
        )}
    ]
    explanation = gemini_call(explain_parts)
    time.sleep(CALL_DELAY)

    return csv_text, explanation or "No explanation generated."


def save_csv(csv_text: str, filepath: str):
    """Parse raw CSV text and save as a clean CSV file."""
    reader = csv.reader(io.StringIO(csv_text))
    rows   = [row for row in reader if any(c.strip() for c in row)]

    if not rows:
        return False

    # Fix blank header cells
    header    = rows[0]
    rows[0]   = [h if h.strip() else f"Col_{i+1}" for i, h in enumerate(header)]

    os.makedirs(os.path.dirname(filepath), exist_ok=True)
    with open(filepath, "w", newline="", encoding="utf-8") as f:
        csv.writer(f).writerows(rows)
    return True


def process_pages(pdf_path: str, pages: list, label: str) -> list:
    """
    Run Gemini extraction on the given list of pages.
    Returns list of result dicts for successfully extracted tables.
    """
    results = []
    for page_num in pages:
        print(f"\n  📄 Page {page_num} ({label}) …")
        csv_text, explanation = extract_table_from_page(pdf_path, page_num)

        if csv_text is None:
            print(f"     ⚠️  No table extracted from page {page_num}.")
            continue

        filename = f"page{page_num}_{label}.csv"
        csv_path = os.path.join(OUTPUT_DIR, filename)
        saved    = save_csv(csv_text, csv_path)

        if saved:
            print(f"     ✅ Saved → {filename}")
            results.append({
                "page":        page_num,
                "label":       label,
                "csv_path":    csv_path,
                "explanation": explanation,
            })
        else:
            print(f"     ⚠️  CSV was empty after parsing, skipped.")

    return results


# ══════════════════════════════════════════════════════
# SAVE REPORT
# ══════════════════════════════════════════════════════

def save_report(results: list, pdf_name: str):
    """Save Markdown report with CSV paths + explanations."""
    report_path = os.path.join(OUTPUT_DIR, "table_explanations.md")

    with open(report_path, "w", encoding="utf-8") as f:
        f.write(f"# Table Explanations — {pdf_name}\n\n")
        f.write(
            "> **Streamlit usage:**\n"
            "> ```python\n"
            "> import pandas as pd, streamlit as st\n"
            "> df = pd.read_csv('extracted_tables/<filename>.csv')\n"
            "> st.dataframe(df, use_container_width=True)\n"
            "> ```\n\n---\n\n"
        )
        for r in results:
            f.write(f"## Page {r['page']} — {r['label']}\n\n")
            f.write(f"**CSV:** `{r['csv_path']}`\n\n")
            f.write(f"### Explanation\n\n{r['explanation']}\n\n---\n\n")

    print(f"\n📄  Report saved → {report_path}")


# ══════════════════════════════════════════════════════
# MAIN
# ══════════════════════════════════════════════════════

def main():
    if GEMINI_API_KEY == "YOUR_API_KEY_HERE":
        print("❌  Set your GEMINI_API_KEY first.")
        print("    export GEMINI_API_KEY='your_key_here'")
        sys.exit(1)

    os.makedirs(OUTPUT_DIR, exist_ok=True)

    doc         = fitz.open(pdf_path)
    total_pages = len(doc)
    doc.close()

    print(f"\n{'═'*50}")
    print(f"  PDF Table Extractor")
    print(f"{'═'*50}")
    print(f"  File  : {pdf_path}")
    print(f"  Pages : {total_pages}\n")

    # ── STEP 1 : Regex detection ───────────────────────
    print("📌  STEP 1 — Regex scan for table captions …\n")
    regex_pages = find_table_pages_regex(pdf_path)
    print(f"  Table pages found by regex: {regex_pages}\n")

    all_results = []

    if regex_pages:
        # ── STEP 2 : Extract from regex pages ─────────
        print(f"{'─'*50}")
        print(f"🤖  STEP 2 — Extracting tables from regex pages via Gemini …")
        all_results = process_pages(pdf_path, regex_pages, "regex")

        # ── STEP 3 : Only if Step 2 found NOTHING ─────
        if not all_results:
            print(f"\n⚠️   Step 2 extracted 0 tables from regex pages.")
            print(f"     Falling back to Gemini scan on all OTHER pages …\n")
            print(f"{'─'*50}")

            other_pages = [p for p in range(1, total_pages + 1) if p not in regex_pages]
            print(f"🤖  STEP 3 — Scanning pages: {other_pages}\n")
            all_results = process_pages(pdf_path, other_pages, "fallback")

    else:
        # Regex found nothing at all → scan all pages
        print("⚠️   Regex found no table captions in this PDF.")
        print(f"     Running Gemini Vision on all {total_pages} pages …\n")
        print(f"{'─'*50}")
        all_pages   = list(range(1, total_pages + 1))
        all_results = process_pages(pdf_path, all_pages, "fallback")

    # ── Save report ────────────────────────────────────
    if all_results:
        save_report(all_results, Path(pdf_path).name)
    else:
        print("\n⚠️   No tables could be extracted from this PDF.")

    print(f"\n{'═'*50}")
    print(f"✅  Done!  Extracted {len(all_results)} table(s).")
    print(f"   CSVs   → ./{OUTPUT_DIR}/*.csv")
    print(f"   Report → ./{OUTPUT_DIR}/table_explanations.md")
    print(f"\n   Streamlit:")
    print(f"   df = pd.read_csv('{OUTPUT_DIR}/<filename>.csv')")
    print(f"   st.dataframe(df, use_container_width=True)")
    print(f"{'═'*50}\n")


if __name__ == "__main__":
    main()
````

````python
OUTPUT_DIR     = Path("extracted_images_1")     
CACHE_DIR      = Path("extracted_assets_1/cache")
MIN_IMAGE_SIZE = 100

def extract_images_from_pdf(pdf_path):
    """
    Extract all meaningful raster images from a PDF using PyMuPDF + Pillow.

    Each image is:
      - saved as a PNG in extracted_images/
      - converted to base64 IN MEMORY (no re-read needed later)

    Returns a list of dicts:
        {page, index, filename, image_path, width, height, b64_data}
    b64_data is used directly for the Gemini API call.
    """
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    doc       = fitz.open(pdf_path)
    extracted = []

    for page_num in range(len(doc)):
        page       = doc[page_num]
        image_list = page.get_images(full=True)

        for img_index, img in enumerate(image_list):
            xref = img[0]

            # Pull raw image bytes from the PDF
            base_image  = doc.extract_image(xref)
            image_bytes = base_image["image"]

            # Open with Pillow to get dimensions and normalise format
            pil_img        = Image.open(io.BytesIO(image_bytes))
            width, height  = pil_img.size

            # Skip decorative images that are too small
            if width < MIN_IMAGE_SIZE or height < MIN_IMAGE_SIZE:
                continue

            # Convert to RGB so everything is a consistent PNG
            if pil_img.mode not in ("RGB", "L"):
                pil_img = pil_img.convert("RGB")

            # Save to disk
            filename  = f"page{page_num + 1}_img{img_index + 1}.png"
            save_path = OUTPUT_DIR / filename
            pil_img.save(str(save_path), "PNG")

            # Encode to base64 from the same in-memory buffer — no file re-read
            buffer = io.BytesIO()
            pil_img.save(buffer, format="PNG")
            b64_data = base64.b64encode(buffer.getvalue()).decode("utf-8")

            extracted.append({
                "page":       page_num + 1,
                "index":      img_index + 1,
                "filename":   filename,
                "image_path": str(save_path),   # consistent with rag/extractor and UI
                "width":      width,
                "height":     height,
                "b64_data":   b64_data,          # ready to send to Gemini directly
            })

    doc.close()
    return extracted
````

````python
def csv_to_markdown(csv_text):
    """
    Convert a raw CSV string returned by Gemini into a Markdown table.
    Escapes pipe characters inside cells so the table renders correctly.
    Returns an empty string if the CSV is empty or invalid.
    """
    reader = csv.reader(io.StringIO(csv_text))
    rows   = [row for row in reader if any(c.strip() for c in row)]

    if not rows:
        return ""

    header = [h.strip() if h.strip() else f"Col_{i+1}" for i, h in enumerate(rows[0])]

    lines = []
    lines.append("| " + " | ".join(header) + " |")
    lines.append("| " + " | ".join(["---"] * len(header)) + " |")

    for row in rows[1:]:
        padded = row + [""] * max(0, len(header) - len(row))
        cells  = [str(c).strip().replace("|", "\\|") for c in padded[:len(header)]]
        lines.append("| " + " | ".join(cells) + " |")

    return "\n".join(lines)
markdown = csv_to_markdown(csv_text)
````
